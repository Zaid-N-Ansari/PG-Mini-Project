```python
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
```

```python
wine = load_wine()
X = pd.DataFrame(data=wine.data, columns=wine.feature_names)

y = pd.Series(wine.target).rename('target').apply(lambda x: wine.target_names[x])

print(f'X shape: {X.shape}\ny shape: {y.shape}\nUnique targets: {wine.target_names}')

data = pd.concat([X,y], axis=1)
data
```

```python
X_scaled = StandardScaler().fit_transform(X)
pd.DataFrame(X_scaled)
```

```python
pca = PCA()
X_pca = pca.fit_transform(X_scaled)
```

```python
X_stdz = pd.DataFrame()
for col in X.columns:
	X_stdz[col] = X[col].apply(lambda x: (x - np.mean(X[col])) / np.std(X[col]))

# are_they_close = np.allclose(X_stdz, X_scaled)
# print(f'Are the manual and X_scaled matrices the same? {are_they_close}')
# for standardizing the data (x - x_bar) / stdev

X_stdz
```

```python
cov_mat = (X_stdz.T @ X_stdz) / (X_stdz.shape[0] - 1) # or X_stdz.cov()
# refer: https://askfilo.com/user-question-answers-smart-solutions/formulas-in-multivariate-statistics-3337333431333038

# are_they_close = np.allclose(cov_mat, X_stdz.cov())
# print(f"Are the manual and np.cov matrices the same? {are_they_close}\n")

cov_mat
```

```python
eigen_vals, eigen_vecs = np.linalg.eig(cov_mat)
eigen_vals.tolist()
```

```python
sorted_idx = np.argsort(eigen_vals)[::-1]
sorted_idx
```

```python
sorted_eigen_vals = eigen_vals[sorted_idx]
sorted_eigen_vecs = eigen_vecs[sorted_idx]
```

```python
explained_var_ratio = sorted_eigen_vals / sorted_eigen_vals.sum()

# are_they_close = np.allclose(explained_var_ratio, pca.explained_variance_ratio_)
# print(f"Are the manual and pca.explained_variance_ratio_ matrices the same? {are_they_close}\n")
# for verification of the manual computation against the built-in fns

explained_var_ratio.tolist()
```

```python
explained_variance = pd.DataFrame({
    'Principal Component': [f'PC{_}' for _ in sorted_idx],
    'Variance Ratio': explained_var_ratio,
    'Cumulative Variance': np.cumsum(explained_var_ratio)
})
explained_variance.index += 1
explained_variance
```

```python
N_COMPONENTS_TO_KEEP = 9

principal_components = sorted_eigen_vecs[0:N_COMPONENTS_TO_KEEP]

X_new = X_stdz @ principal_components.T

print(f"Original data shape: {X_scaled.shape}")
print(f"New transformed data shape: {X_new.shape}")
```
